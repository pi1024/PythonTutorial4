# World Bank Internet Usage Analysis

```python
import wbdata
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# prepare data
indicators = {
    "NY.GDP.PCAP.KD": "gdp_per_capita",
    "NY.GDP.PCAP.KD.ZG": "gdp_growth",
    "SP.DYN.LE00.IN": "life_expectancy",
    "SP.POP.TOTL": "population",
    "EG.FEC.RNEW.ZS" : 'renewble',
    "SE.TER.ENRR" : 'education',
    'IT.NET.USER.ZS' : 'internet',
}

countries = ["GBR", "FRA", "ESP", "ITA",
             "NLD", "CHN", "USA", 'JPN',
             "KOR", "CHE", "VNM", "ARG"]

# retriving the desired data
df_hw = wbdata.get_dataframe(indicators,
                          country=countries,
                          date=("2000", "2025"))




df_hw = df_hw.reset_index()
df_hw = df_hw.rename(columns={"date": "year"})
df_hw["year"] = df_hw["year"].astype(int)

info = wbdata.get_countries(countries)
region_of = {c["name"]: c["region"]["value"] for c in info}
df_hw["region"] = df_hw["country"].map(region_of)


#display data style
print(df_hw.shape)
df_hw.info()
df_hw.head(5)
```

```python
# prepare data
gdp_internet = df_hw[df_hw['year'] == 2024]

#graph
fig, ax = plt.subplots(figsize=(9, 6))
#regression & scatter
sns.regplot(data = gdp_internet, x = "gdp_per_capita", y = "internet",
            scatter_kws={"s": 100, "alpha": 0.8},
            line_kws={"color": "orange", "linewidth": 2},
            ci= 95,
            ax=ax)

ax.set_xscale("log")
ax.set_xlabel("GDP per capita (constant 2015 US$, log scale)")
ax.set_ylabel("Individuals using the internet (%)")
ax.set_title("Higher internet use may accompany higher GDP per capita (2024 )",
             loc="left", fontsize=14, fontweight="bold", pad=12)

ax.grid(alpha=0.25)

# country name
for _, row in gdp_internet.iterrows(): # _ index & row each row data
    ax.annotate(row["country"], (row["gdp_per_capita"], row["internet"]),
                xytext=(5, 5),
                textcoords="offset points",
                fontsize=10)

fig.text(0.01, 0.01, "Source: World Bank (ITU) | Author: Wren",
         ha="left", fontsize=9, style="italic")
fig.tight_layout(rect=[0, 0.04, 1, 1])
g.savefig("/content/drive/MyDrive/Result/analysis1_internet_vs_gdp.png", dpi=300, bbox_inches="tight")
plt.show()

# correlation
print("Pearson correlation:", round(gdp_internet["internet"].corr(gdp_internet["gdp_per_capita"]), 4))
print("Spearman correlation:", round(gdp_internet["internet"].corr(gdp_internet["gdp_per_capita"], method="spearman"), 4))
```

```python
# prepare data
life_internet = df_hw[df_hw['year'] == 2024]



#graph
fig, ax = plt.subplots(figsize=(9, 6))
#regression & scatter
sns.regplot(data = life_internet, x = "internet", y = "life_expectancy",
            scatter_kws={"s": 100, "alpha": 0.8},
            line_kws={"color": "orange", "linewidth": 2},
            ci= 95,
            ax=ax)


ax.set_xlabel("Individuals using the internet (% of population)")
ax.set_ylabel("Life expectancy at birth (years)")
ax.set_title("Higher internet use may accompany longer life expectancy (2024)",
             loc="left", fontsize=14, fontweight="bold", pad=12)

ax.grid(alpha=0.25)

# country name
for _, row in life_internet.iterrows(): # _ index & row each row data
    ax.annotate(row["country"], (row["internet"], row["life_expectancy"]),
                xytext=(5, 5),
                textcoords="offset points",
                fontsize=10)

fig.text(0.01, 0.01, "Source: World Bank (ITU) | Author: Wren",
         ha="left", fontsize=9, style="italic")
fig.tight_layout(rect=[0, 0.04, 1, 1])
g.savefig("/content/drive/MyDrive/Result/analysis2_internet_vs_life_expectancy.png", dpi=300, bbox_inches="tight")
plt.show()


print("Pearson correlation:", round(life_internet["internet"].corr(life_internet["life_expectancy"]), 4))
print("Spearman correlation:", round(life_internet["internet"].corr(life_internet["life_expectancy"], method="spearman"), 4))
```

```python
#prepare data
internet = df_hw.loc[df_hw["year"] == 2024, ["country", "internet"]].sort_values("internet")

# graph
fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(internet["country"], internet["internet"], color="#FFB69F")
ax.set_xlim(0, 105)
ax.set_xlabel("Individuals using the internet (% of population)")
ax.set_ylabel("Country")
ax.set_title("Internet-use rates differ across countries (2024)",
             loc="left", fontsize=14, fontweight="bold", pad=12)
ax.grid(axis="x", alpha=0.25)
ax.grid(axis="y", visible=False)

top = internet.iloc[-1]
bottom = internet.iloc[0]

# point out highest & lowest
ax.annotate(f"Highest: {top['country']} ({top['internet']:.1f}%)",
            xy=(top["internet"], top["country"]),
            xytext=(-8, -3), textcoords="offset points", color = "white",
            ha="right", va="bottom", fontsize=12)

ax.annotate(f"Lowest: {bottom['country']} ({bottom['internet']:.1f}%)",
            xy=(bottom["internet"], bottom["country"]),
            xytext=(-8, -3), textcoords="offset points", color = "white",
            ha="right", va="bottom", fontsize=12)

fig.text(0.01, 0.01, "Source: World Bank (ITU) | Author: Wren",
         ha="left", fontsize=9, style="italic")
fig.tight_layout(rect=[0, 0.04, 1, 1])
fig.savefig("/content/drive/MyDrive/Result/analysis3_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

print("Comparison year:", internet.sort_values("internet", ascending=False))
```
